```python
import requests
import pandas as pd
from getpass import getpass
from datetime import datetime
from zoneinfo import ZoneInfo

# API configuration
API_KEY = getpass("431c3b826f77794a2649b70e022e6914")
CITY = "Indore"
COUNTRY = "IN"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": f"{CITY},{COUNTRY}",
    "appid": API_KEY,
    "units": "metric"
}

# Fetch weather data
response = requests.get(url, params=params, timeout=20)
response.raise_for_status()
data = response.json()

# Convert timestamps to Indian Standard Time
ist = ZoneInfo("Asia/Kolkata")
sunrise = datetime.fromtimestamp(
    data["sys"]["sunrise"], ist
).strftime("%Y-%m-%d %H:%M:%S")

sunset = datetime.fromtimestamp(
    data["sys"]["sunset"], ist
).strftime("%Y-%m-%d %H:%M:%S")

# Create weather report
weather_report = {
    "City": data["name"],
    "Country": data["sys"]["country"],
    "Temperature (°C)": data["main"]["temp"],
    "Feels Like (°C)": data["main"]["feels_like"],
    "Humidity (%)": data["main"]["humidity"],
    "Pressure (hPa)": data["main"]["pressure"],
    "Weather": data["weather"][0]["main"],
    "Description": data["weather"][0]["description"],
    "Wind Speed (m/s)": data["wind"]["speed"],
    "Cloudiness (%)": data["clouds"]["all"],
    "Sunrise (IST)": sunrise,
    "Sunset (IST)": sunset,
    "Fetched At (IST)": datetime.now(ist).strftime(
        "%Y-%m-%d %H:%M:%S"
    )
}

df = pd.DataFrame([weather_report])

print("WEATHER REPORT")
display(df)

# Save the report
df.to_csv("weather_report.csv", index=False)
print("\nWeather report saved as weather_report.csv")
```
