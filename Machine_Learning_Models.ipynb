# Machine Learning Models

## Aim
To study different Machine Learning models, their working principles, types, and applications.

## Introduction
Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

Machine Learning models learn from training data and are evaluated using unseen test data.

## Main Types of Machine Learning
1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning


## 1. Linear Regression
Linear Regression predicts continuous numerical values by finding a relationship between input variables and an output variable.

Example: Predicting house prices.

## 2. Logistic Regression
Logistic Regression estimates the probability of a class and is commonly used for classification problems.

Example: Spam or ham message classification.

## 3. Decision Tree
A Decision Tree makes predictions by splitting data using conditions arranged in a tree-like structure.

Example: Predicting whether a customer will buy a product.

## 4. Random Forest
Random Forest combines predictions from multiple decision trees to improve reliability and reduce overfitting.

Example: Fraud detection and classification.

## 5. Support Vector Machine (SVM)
SVM finds a decision boundary that separates different classes, aiming to maximize the margin between them.

Example: Text classification and image recognition.

## 6. K-Nearest Neighbors (KNN)
KNN predicts a class or numerical value using nearby data points.

Example: Classifying similar items based on their features.

## 7. Naive Bayes
Naive Bayes is a probabilistic classifier based on Bayes' theorem. It assumes that features are conditionally independent given the class.

Example: SMS spam detection.

## 8. K-Means Clustering
K-Means is an unsupervised learning algorithm that groups data points into K clusters based on similarity.

Example: Customer segmentation.

## 9. Artificial Neural Network (ANN)
An ANN consists of interconnected artificial neurons arranged in layers. It learns complex relationships from data.

Example: Image and speech recognition.

## 10. Reinforcement Learning
Reinforcement Learning trains an agent to select actions by receiving rewards or penalties from an environment.

Example: Game-playing agents and robotics.

## Comparison of Machine Learning Models

| Model | Learning Type | Main Use |
|---|---|---|
| Linear Regression | Supervised | Numerical prediction |
| Logistic Regression | Supervised | Classification |
| Decision Tree | Supervised | Classification and regression |
| Random Forest | Supervised | Classification and regression |
| SVM | Supervised | Classification and regression |
| KNN | Supervised | Classification and regression |
| Naive Bayes | Supervised | Probabilistic classification |
| K-Means | Unsupervised | Clustering |
| ANN | Commonly supervised; can vary | Complex pattern learning |
| Reinforcement Learning | Reward-based learning | Sequential decision-making |

## Conclusion
Different Machine Learning models solve different types of problems. The choice of model depends on the dataset, task, performance requirements, and interpretability.
